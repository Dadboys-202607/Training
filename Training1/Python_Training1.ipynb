```python
import pandas as pd

df = pd.read_excel("sales.xlsx")

# 合計金額列を object 型にして混在を許容
df["合計金額"] = df["合計金額"].astype("object")

# 2行目以降だけ計算
df.loc[1:, "合計金額"] = df["数量"][1:] * df["単価"][1:]

print(df)

# 商品別集計
df_sum = df.groupby("商品名")[["数量", "合計金額"]].sum().reset_index()

print(df_sum)

df_sum.to_excel("output.xlsx", index=False)
```
